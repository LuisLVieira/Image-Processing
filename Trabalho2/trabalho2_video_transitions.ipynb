# Detecção de Transições em Vídeos

## Diferença entre pixels

```python
video = 'videos/indi009.mpg'


videos = utils.Videos(video)

indexes, measures = videos.pixel_diff(T1 = 50, T2=25000)

videos.plot_measures(measures, title = "Diferença entre pixels T1 50 T2 25000", T=25000, output_path='outputs')

summary = videos.save_video(indexes, title = "pixel_diff_T1_50_T2_25000", output_path="outputs/videos")

videos.plot_summary_images(summary, "Imagens Diferença entre Pixels T1=50 T2=25000")
```

```python
videos = utils.Videos('videos/umn.mp4')


indexes, measures = videos.pixel_diff(T1 = 50, T2=1500)


videos.plot_measures(measures, title = "Diferença entre pixels T1 50 T2 1500", T=1500, output_path='outputs')

summary = videos.save_video(indexes, title = "pixel_diff_T1_50_T2_1500", output_path="outputs/videos")

videos.plot_summary_images(summary, "Imagens Diferença entre Pixels T1=50 T2=25000")
```

## Diferença entre blocos

```python
video = 'videos/umn.mp4'


videos = utils.Videos(video)

indexes, measures = videos.block_diff(T1 = 800, T2=4, n_blocks=8)

videos.plot_measures(measures, title = "Diferença entre blocos T1 800 T2 4 8x8", T=4, output_path='outputs')

summary = videos.save_video(indexes, title = "block_diff_T1_800_T2_4", output_path="outputs/videos")

videos.plot_summary_images(summary, "Imagens Diferença entre Blocos T1=500 T2=4")
```
