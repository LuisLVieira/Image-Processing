```python
import glob
import os
from utils import (
    get_fft_spectrum,
    plot_spectrum,
    get_spectrum_energy_histogram,
    plot_spectrum_energy_histogram
)
```

## Visualização das imagens no domínio da frequência

```python
for image_path in glob.glob('images/*'):

    img, fft_mag, fft_mag_cent = get_fft_spectrum(image_path)
    plot_spectrum(
        img,
        fft_mag,
        fft_mag_cent,
        title = os.path.basename(image_path),
        output_path='./outputs'
)
```

## Obtenção dos histogramas de energia por ângulo

```python
for image_path in glob.glob('images/*'):

    img, _, fft_mag_cent = get_fft_spectrum(image_path)

    print(fft_mag_cent.shape)
    hist, bin_centers = get_spectrum_energy_histogram(
        spectrum = fft_mag_cent,
        bins = 18,
        exclude_ratio = 10
    )

    plot_spectrum_energy_histogram(
        image=img,
        spectrum=fft_mag_cent,
        bin_centers=bin_centers,
        hist=hist,
        title = os.path.basename(image_path)
)
```

```python

image_path = './images/exemplo1.png'

img, _, fft_mag_cent = get_fft_spectrum(image_path)

hist, bin_centers = get_spectrum_energy_histogram(
    spectrum = fft_mag_cent,
    bins = 36,
    exclude_ratio = 10
)

plot_spectrum_energy_histogram(
    image=img,
    spectrum=fft_mag_cent,
    bin_centers=bin_centers,
    hist=hist,
    title = os.path.basename(image_path)
)
```
