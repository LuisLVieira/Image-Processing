## Visualização das imagens no domínio da frequência

- É possível perceber que para todas as imagens é necessário centralizar o espectro para as baixas frequências estarem presentes no centro da imagem (coordenada 0).

```python
for image_path in glob.glob('images/*'):
    img = cv2.imread(image_path)

    img, fft_mag, fft_mag_cent = get_fft_spectrum(img)
    plot_spectrum(
        img,
        fft_mag,
        fft_mag_cent,
        title = f"{os.path.basename(image_path).replace('.png', '').replace('.pgm', '')}_centralização",
        output_path='./outputs',
        show_raw_spectrum = True
    )
```

## Testes de invariância com relação à transformações geométricas

- Foi observado que os ângulos dominates não se alteram na maioria dos casos para a rotação, o que era esperado, pois é nítido o deslocamento do espectro no mesmo ângulo da imagem. Este fenômeno possivelmente foi causado por descontinuidades entre bordas opostas, gerando componentes espectrais artificiais e produzindo ângulos fixos (fenômeno de Gibbs).Uma solução seria a suavização ante de aplicar a DFT, aplicando janela de Hann (mais eficaz), janela Gaussiana ou subtração do valor médio.

```python
image_path = './images/monarch.pgm'

img = cv2.imread(image_path)

plots(
    img,
    title=os.path.basename(image_path),
    output_path = './outputs',
    save_img=False
)

image_translated = translate_image(img, 50, 50)

plots(
    image_translated,
    title=f'{os.path.basename(image_path)}_transladado_50p',
    output_path = './outputs',
    save_img=True
)

image_rotated = rotate_image(img, 45)

plots(
    image_rotated,
    title=f'{os.path.basename(image_path)}_rotacionado_45',
    output_path = './outputs',
    save_img=True
)

image_scaled = scale_image(img, 0.5)

plots(
    image_scaled,
    title=f'{os.path.basename(image_path)}_reduzida_0,5x',
    output_path = './outputs',
    save_img=True
)

image_scaled = scale_image(img, 1.5)

plots(
    image_scaled,
    title=f'{os.path.basename(image_path)}_ampliada_1,5x',
    output_path = './outputs',
    save_img=True
)
```
